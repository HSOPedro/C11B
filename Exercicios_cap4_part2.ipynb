Baseado nos commandos que vimos até
o momento
e no Dataset
fornecido
, crie scripts em Python que respondam às seguintes
perguntas
:
1. Apresente
a porcentagem de missões que deram certo
2. Qual
a media de gastos de uma missão especial se baseando em
missões que possuam valores disponíveis (>
0)?
3. Encontre quantas missões espaciais neste Dataset foram realizadas
pelos Estados Unidos (EUA)
4. Encontre qual foi
a missão mais cara realizada pela empresas
“SpaceX
”
5. Mostre
o nome das empresas que já realizaram missões espaciais,
juntamente com suas respectivas quantidades de missões (use
o for
no final para mostrar as informações)


```python
import numpy as np

dataset = np.loadtxt('space-1.csv', delimiter=';', dtype= "str")



statusmissao = dataset[1:, -1]
successcount = np.sum(statusmissao == 'Success')
totalmissi = len(statusmissao)

percensucce = (successcount / totalmissi) * 100

print(f"Porcentagem de missões que deram certo: {percensucce:.2f}%")
print()


gastos = dataset[1:, -2]

gastos = gastos[gastos != '']

gastos = gastos.astype(float)

gastosvalidos = gastos[gastos > 0]

mediagastos = np.mean(gastosvalidos)

print(f"Média de gastos das missões: {mediagastos:.2f} milhões")
print()


localizacoes = dataset[1:, 2]

contagemusa = np.sum(np.char.endswith(localizacoes, 'USA'))

print(f"Quantidade de missões realizadas pelos EUA: {contagemusa}")
print()

empresas = dataset[1:, 1]
missoes = dataset[1:, 4]
gastos = dataset[1:, 6]

filtro = (empresas == 'SpaceX') & (gastos != '')

missoes_spacex = missoes[filtro]
gastos_spacex = gastos[filtro].astype(float)

maiorgasto = np.argmax(gastos_spacex)

print(f"Missão mais cara da SpaceX: {missoes_spacex[maiorgasto]}")
print(f"Valor da missão: {gastos_spacex[maiorgasto]:.2f} milhões")
print()



empresas = dataset[1:, 1]

empresasunicas, quantidades = np.unique(empresas, return_counts=True)

for i in range(len(empresasunicas)):
    print(f"{empresasunicas[i]}: {quantidades[i]} missões")
```

Baseado nos commandos que vimos até

, crie scripts em Python que respondam às seguintes
perguntas
:
- 6. Qual
a porcentagem de missões realizadas com foguetes cujo status
é
"StatusRetired
" (coluna Status Rocket)?
- 7. Quantas missões foram lançadas
a partir de localizações que contêm
"Russia
" (coluna Location)?
- 8. Encontre
a empresa
e
o valor da missão mais cara de todo

.

```python
import numpy as np

dataset = np.loadtxt('space-1.csv', delimiter=';', dtype="str")

statusfoguete = dataset[1:, -3]

retiredcount = np.sum(statusfoguete == 'StatusRetired')

totalmissi = len(statusfoguete)

percentretired = (retiredcount / totalmissi) * 100

print(f"Porcentagem de missões com foguetes aposentados: {percentretired:.2f}%")
print()

localizacoes = dataset[1:, 2]

contagemrussia = np.sum(np.char.find(localizacoes, 'Russia') >= 0)

print(f"Quantidade de missões lançadas a partir da Rússia: {contagemrussia}")
print()

empresas = dataset[1:, 0]
gastos = dataset[1:, -2]

gastosvalidos = gastos[gastos != '']

gastosvalidos = gastosvalidos.astype(float)

maiorgasto = np.max(gastosvalidos)

posicao = np.where(gastos.astype(float) == maiorgasto)[0][0]

print(f"Empresa: {empresas[posicao]}")
print(f"Valor da missão: {maiorgasto:.2f} milhões")
print()

empresas = dataset[1:, 0]
gastos = dataset[1:, -2]

filtro = gastos != ''

empresas = empresas[filtro]
gastos = gastos[filtro].astype(float)

maiorgasto = np.max(gastos)

posicao = np.argmax(gastos)

print(f"Empresa: {empresas[posicao]}")
print(f"Valor da missão: {maiorgasto:.2f} milhões")
print()
```
