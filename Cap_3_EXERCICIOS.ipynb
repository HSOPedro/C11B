4. Faça um programa que leia o nome e peso de
3 pessoas
e no final mostre
o
nome da pessoa mais pesada
e
a mais leve
;

```python
pessoas = []

for i in range(3):
    nome = input(f'Digite o nome da {i + 1}ª pessoa: ')
    peso = float(input(f'Digite o peso da {i + 1}ª pessoa: '))
    pessoas.append((peso, nome))

pessoas.sort()

mais_leve = pessoas[0]
mais_pesada = pessoas[-1]

print(f'\nMais pesada: {mais_pesada[1]}')
print(f'Mais leve: {mais_leve[1]}')
```

5
. Desenvolva um programa que leia
o nome, idade
e sexo de
n pessoas
. No
final, mostre
:
a.
- A média de idade do grupo
;
- b. Quantas mulheres têm menos de 20 anos
.
- Dica
: em Python, os operadores booleanos básicos são and
, or
e not


```python
quantidade = int(input('Digite a quantidade de pessoas: '))

pessoas = []

for i in range(quantidade):
    nome = input(f'\nNome da {i + 1}ª pessoa: ')
    idade = int(input(f'Idade da {i + 1}ª pessoa: '))
    sexo = input(f'Sexo da {i + 1}ª pessoa (H ou M): ')

    pessoas.append((nome, idade, sexo))

soma_idades = 0
mulheres_jovens = 0

for pessoa in pessoas:
    soma_idades += pessoa[1]

    if pessoa[2] == 'M' and pessoa[1] < 20:
        mulheres_jovens += 1

media = soma_idades / quantidade

print(f'\nMedia de idade: {media}')
print(f'Mulheres com menos de 20 anos: {mulheres_jovens}')
```

. Crie um dicionário para armazenar os dados de um produto (nome, preço
e
quantidade em estoque)
. Peça ao usuário os dados de
3 produtos diferentes
e
guarde cada dicionário em uma lista
. No final, percorra
a lista
e mostre, para
cada produto, seu nome
e
o valor total em estoque (preço
× quantidade)
.

```python
produtos = []

for i in range(3):
    nome = input('\nNome do produto: ')
    preco = float(input('Preco: R$ '))
    quantidade = int(input('Quantidade em estoque: '))

    produto = {
        'nome': nome,
        'preco': preco,
        'quantidade': quantidade
    }

    produtos.append(produto)

for item in produtos:
    total = item['preco'] * item['quantidade']
    print(f"\n{item['nome']}: R$ {total:.2f}")
```
