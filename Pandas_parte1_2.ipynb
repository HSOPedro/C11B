- 6. Utilizando do DataFrame exemplo do tópico 5.3 deste material, calcule a
média dos elementos da coluna X que são menores que 30;
- 7. Utilizando do mesmo DataFrame, apresente a média dos elementos da
linha D usando a função loc() como base e a soma dos elementos da linha E
usando a função iloc() como base;
- 8. Faça um Slicing na matriz mostrando apenas as linhas A, C e E
juntamente com as colunas X e Y. Em seguida, mostre qual seria a soma dos
elementos de cada uma destas linhas e cada uma destas colunas.

```python
##6

import numpy as np

# definindo uma semente para os números aleatórios
np.random.seed(10)

# montando o DataFrame
dados = np.random.randint(1, 50, (5, 4))

df = pd.DataFrame(dados,index=['A', 'B', 'C', 'D', 'E'],columns=['W', 'X', 'Y', 'Z'])

print(df)

print()

print("Media dos elementos em X menores que 30 : ")
print(df[df['X'] < 30]['X'].mean())
```

```python
##7

# usando o loc na linha D
print("Media dos elementos da linha D :")
print(df.loc['D'].mean())

# usando o iloc na linha E
print("\nSoma dos elementos da linha E :")
print(df.iloc[4].sum())
```

```python
##8
slicing = df.loc[['A', 'C', 'E'], ['X', 'Y']]
print("Slicing da Matriz :")
print(slicing)

print('\n')

for i in range(len(slicing)):
  print(f"Soma dos elementos da {i}ª linha : {slicing.iloc[i].sum()}")

print('\n')

for i in range(len(slicing.columns)):
  print(f"Soma dos elementos da {i}ª coluna : {slicing.iloc[:,i].sum()}")
```
