```python

#impot das bibliotecas
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


#valores de x e y
x = np.array([1, 2, 3, 4])
y = x*2

#eixo y 2

y2 = x**2

#legendas x y
plt.xlabel('valores de x')
plt.ylabel("valores de y")

#titulo
plt.title("grafico de linhas")

plt.plot(x, y, "*:r", x, y2, "s--b" ,linewidth=3, markersize=20)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#valores
x = np.array([1, 2, 3, 4])
y = x*2

y2 = x**2

#legendas
plt.xlabel('valores de x')
plt.ylabel("valores de y")

plt.title("grafico de linhas")

plt.subplot(1,2,1)
plt.plot(x, y, "*:r",linewidth=3, markersize=20)
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(x, y2, "s--b" ,linewidth=3, markersize=20)
plt.grid(True)

plt.show()
```

```python
ds = pd.read_csv("paises.csv", delimiter=";")

ds.head()

#maiores paisses
maiores = ds.nlargest(6, 'Area (sq. mi.)')

print(maiores)

#tracando o grafico
plt.scatter(maiores['Country'], maiores['GDP ($ per capita)'],s=maiores['Area (sq. mi.)']/10000)
plt.show()

```

```python
maioresGPD = ds.nlargest(5, 'GDP ($ per capita)')

print(maioresGPD)

#garfico

plt.bar(maioresGPD['Country'], maioresGPD['GDP ($ per capita)'])
plt.show()
```

```python
paisessemcosta = ds[ds['Coastline (coast/area ratio)'] == 0]

paisessemcosta

qtsemcosta = len(paisessemcosta.count())

tqcomcosta = len(ds) - qtsemcosta

plt.pie(x=[qtsemcosta, tqcomcosta], labels=['sem costa', 'com costa'], autopct='%1.1f%%')
plt.show()
```
