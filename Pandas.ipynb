Pandas
- series (1 - d)

```python
import numpy as np
import pandas as pd


# conjunto de labels
# conjunti de valores

labels = ['pedro', 'lara', 'lion', 'gigi',]

valores = [21, 20, 23, 19]

#criando sries

se1 = pd.Series(data=valores, index=labels)

print(se1)
print(type(se1))

#acessando elementos da series

print(se1['lara'])
print(se1[['lara', 'pedro']])
```

```python
s1 = pd.Series({'v1':10,'v2':45,'v3':30})

s2 = pd.Series({'v1':10,'v4':50,'v3':70})

print(s1,s2)
print()
print(s1 + s2)
print()
print(s1.add(s2, fill_value=0))
```

Padas
- datafreme (2d)

```python
import numpy as np
import pandas as pd

#como prenecher um datafreme
# lista de labels (colunas)
colunas = ['W', 'X', 'Y', 'Z']
#lista de labels (linhas)
linhas = ['A', 'B', 'C', 'D', 'E']

np.random.seed(10)
#lista de valores
valores = np.random.randint(1,50,[5,4])

#criando dataframe

df = pd.DataFrame(data=valores, index=linhas, columns=colunas)

print(df)
```

Manipilando o datafreme

```python
print(df)

print()

#unica coluna

print(df['X'])

print()

#puxando 2 colunas

print(df[['Y', 'Z']])

print()

#unica celula

print(df['Y']['C'])

print()

#puxando multiplas celulas

print(df[['W','X',"Z"]])

print()

```

slincin de dados no dataframe com loc(labels) e iloc(indices de labels)

```python
import numpy as np
import pandas as pd

print(df)
print()

#puxando uma unica linha do dataframe

print(df.loc['A', ['W','X','Y','Z',]])

print()

#puxando multiplas linhas do dataframe

print(df.loc[['B','E'], ['W','X','Y','Z',]])

print()

#dois por dois

print(df.loc[['D','E'], ['Y','Z']])

print()

#usando o iloc

print(df.iloc[2, :])

print()

print(df.iloc[[1,4],:])

print()

print(df.iloc[0:2, 0:2])
```
