```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
dados_air = pd.read_csv(
    'airtravel.csv',
    delimiter=',',
    index_col='Date',
    parse_dates=True
)

dados_air['Passengers'].plot(
    figsize=(8, 6),
    title='Quantidade de passageiros ao longo do tempo',
    xlabel='Data',
    ylabel='Número de passageiros',
    x_compat=True
)

plt.show()
```

```python
resultado_air = seasonal_decompose(
    dados_air['Passengers'],
    model='additive'
)

resultado_air.plot()
plt.tight_layout()
plt.show()
```

```python

#Tendência: A série apresenta uma tendência crescente, indicando que a quantidade de passageiros aumenta com o passar do tempo.



#Sazonalidade: Existe um comportamento sazonal, com picos que se repetem aproximadamente no mesmo período de cada ano, principalmente próximo ao mês 9.



#Ciclo: Não é possível identificar claramente um ciclo, pois o intervalo de dados não é suficiente para observar variações de longo prazo.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
dados_co2 = pd.read_csv(
    'co2_emissions.csv',
    delimiter=',',
    index_col='Year',
    parse_dates=True
)

dados_co2['CO2_Emissions'].plot(
    figsize=(8, 6),
    title='Emissão de CO2 ao longo dos anos',
    xlabel='Ano',
    ylabel='Emissão de CO2',
    x_compat=True
)

plt.show()
```

```python
resultado_co2 = seasonal_decompose(
    dados_co2['CO2_Emissions'],
    model='additive'
)

resultado_co2.plot()
plt.tight_layout()
plt.show()
```

```python
#Tendência: A série apresenta uma tendência decrescente, mostrando uma redução nos valores ao longo do período analisado.



#Sazonalidade: Não apresenta sazonalidade identificável, pois não existe um padrão de repetição em intervalos regulares.



#Ciclo: É possível observar um comportamento cíclico, já que os valores apresentam variações ao longo de períodos maiores.
```
