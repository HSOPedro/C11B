####2. Mostre a tabuada de um número que o usuário escolher dentro de um intervalo específico também escolhido por ele


```python
numero = int(input('Digite um número: '))
intervalo1 = int(input('Digite um intervalo inicial: '))
intervalo2 = int(input('Digite outro intervalo final: '))

for i in range(intervalo1, intervalo2 + 1):
  print(numero, 'x', i, '=', numero * i)
```

3. Faça um programa que leia o sexo de uma pessoa e diga se ela é
homem (caso seja digitado M) ou mulher (caso seja digitado F). Caso
seja digitado algo inválido, continue perguntando até que o usuário
entre com um sexo válido

```python
sexo = input("Qual é o sexo (M = Homem) (F = Mulher) ?" )

while(sexo != 'M' and sexo != 'F'):
  sexo = input("Qual é o sexo (M = Homem) (F = Mulher) ?")
if sexo == 'M':
  print('Homem')
elif sexo == 'F':
  print('Mulher')
else:
  print('Sexo inválido')
```

4. Desenvolva um script que pergunte a distância de uma viagem em
Km. Calcule o preço da passagem, cobrando 0.50   por Km para viagens até 200Km e 0.45 para viagens mais longas negrito

```python
distacia = int(input('Qual a distância da viagem? '))
if distacia <= 200:
  print('o valor é R$' , distacia * 0.50)
else:
  print('o valor é R$', distacia * 0.45)
```

5. Faça um programa que leia um número entre 1000 e 9999 e mostre na tela
  
    • qual o número da unidade
  
    • número da dezena
  
    • número da centena
  
    • E número do milhar
  

```python
numero = (input('Digite um número entre 1000 e 9999: '))
print('Unidade: ', numero[3:])
print('Dezena: ', numero[2:3])
print('Centena: ', numero[1:2])
print('Milhar: ', numero[:1])
```

6. Peça ao usuário para entrar com um número decimal. Em seguida,
aplique e mostre o resultado:

    • da raiz quadrada deste número
    
    • função teto
    
    • função chão
    
    • sua parte inteira

```python
import math
numero = float(input('Digite um número decimal: '))
print('Raiz quadrada: ', math.sqrt(numero))
print('Funcao Teto: ',math.ceil(numero))
print('funcao Chão: ',math.floor(numero))
print('Parte inteira: ',int(numero))
```

7. Faça um programa que leia uma palavra. Esse programa deve
percorrer a palavra e imprimir cada letra em maiúsculo, uma letra por
linha. No final, deverá informar quantas vogais a palavra tem e se a
letra ‘A’ está presente nela

```python
palavra = input('Digite uma plavra:  ')
palavra.split()
print(palavra.upper())
cont = 0
vogais = {'a','e','i','o','u'}
for letra in palavra:
  if letra in vogais:
    cont += 1
if 'a' in palavra:
    print('A letra A está presente')
else:
    print('A letra A não está presente')

print('A palavra tem', cont, 'vogais')
```

8. Crie um programa que leia dois números calcule e mostre na tela:
    • Adição
    
    • Subtração

    • Multiplicação

    • Divisão

    • Resto da divisão

    • Potência (primeiro número elevado ao segundo)

```python
numero = int(input('Digite o prieiro número: '))
numero2 = int(input('Digite o segundo número: '))
print('Adição: ', numero + numero2)
print('Subtração: ', numero - numero2)
print('Multiplicação: ', numero * numero2)
print('Divisão: ', numero / numero2)
print('Resto da divisão: ', numero % numero2)
print('Potência: ', numero ** numero2)
```
