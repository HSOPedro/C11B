Questão 1  
- Crie um programa que leia o nome e ano de várias músicas (até que
o usuário não queira mais cadastrar), guardando os dados de cada
música em um dicionário e todos os dicionários em uma lista. No
final, mostre:  
- a. Quantas músicas foram cadastradas;  
- b. As Informações da(s) música(s) do ano mais antigo;

Para esta questão, faça o que se pede em ordem:  
- a. Crie dois NumPy Arrays 1-D com 4 nomes de pessoas cada;  
- b. Em seguida, concatene-os em um só Array;  
- c. Transforme o Array final em um Array 2-D com mais colunas
do que linhas;  
- d. Por fim, ordene os nomes do Array 2-D em ordem
decrescente;


```python
import numpy as np

nomes_a = np.array(["Lala", "caxambu", "Gigi", "hiro"])
nomes_b = np.array(["Dudao", "Thais", "Nath", "Sara"])

todos_nomes = np.concatenate([nomes_a, nomes_b])

print("Nomes reunidos:")
print(todos_nomes)

tabela = todos_nomes.reshape(2, 4)

print("\nMatriz original:")
print(tabela)

tabela_ordenada = np.sort(tabela, axis=1)[:, ::-1]

print("\nMatriz em ordem decrescente:")
print(tabela_ordenada)
```

Questão 3
- Os valores de cores RGBA são uma extensão dos valores de cores
RGB com um canal alfa que especifica a opacidade de uma cor. Um
valor de cor RGBA é especificado com:  
rgba (vermelho, verde, azul, alfa). O parâmetro alfa é um número
entre 0,0 (totalmente transparente) e 1,0 (totalmente opaco).
Baseado nestas informações e na coleção abaixo, crie códigos e
imprima os resultados que respondam às seguintes perguntas:


- colors = [
{"color": "black", "type": "primary", "code": {"rgba": [255,255,255,1],"hex": "#000"}},
{"color": "green", "type": "secondary", "code": {"rgba": [0,255,0,0.1],"hex": "#0F0"}},
{"color": "yellow", "type": "primary","code": {"rgba": [255,255,0,0.7],"hex": "#FF0"}},
{"color": "blue", "type": "primary","code": {"rgba": [0,0,255,1],"hex": "#00F"}}
]


- a. Mostre apenas o nome das cores que são primárias;
- b. Mostre apenas os códigos hexadecimais das cores que possuem tom de
azul máximo (255);
- c. A partir da coleção colors, extraia seus valores para criar um NumPy
Array 1-D formado apenas pelo nome e código hexadecimal de cada
cor;
Ex: “black”, ”#000”, “green”, “#0F0”...
- d. Transforme o Array 1-D em um Array 2-D no seguinte padrão:
...
green #0F0
yellow #FF0
...
- e. Troque o nome de cada cor no Array 2-D para seu respectivo nome em
Português;

```python
import numpy as np

colors = [
    {"color": "red", "type": "primary", "rgba": [255, 0, 0, 1], "hex": "#FF0000"},
    {"color": "green", "type": "secondary", "rgba": [0, 255, 0, 1], "hex": "#00FF00"},
    {"color": "blue", "type": "primary", "rgba": [0, 0, 255, 1], "hex": "#0000FF"},
    {"color": "yellow", "type": "primary", "rgba": [255, 255, 0, 1], "hex": "#FFFF00"},
    {"color": "cyan", "type": "secondary", "rgba": [0, 255, 255, 1], "hex": "#00FFFF"},
    {"color": "magenta", "type": "secondary", "rgba": [255, 0, 255, 1], "hex": "#FF00FF"},
    {"color": "black", "type": "other", "rgba": [0, 0, 0, 1], "hex": "#000000"},
    {"color": "white", "type": "other", "rgba": [255, 255, 255, 1], "hex": "#FFFFFF"}
]

lista_primarias = []

for cor in colors:
    if cor["type"] == "primary":
        lista_primarias.append(cor["color"])

print("Cores primárias:")
print(lista_primarias)


hex_azul = []

for cor in colors:
    if cor["rgba"][2] == 255:
        hex_azul.append(cor["hex"])

print("\nCores com canal azul igual a 255:")
print(hex_azul)


dados = []

for cor in colors:
    dados.append(cor["color"])
    dados.append(cor["hex"])

array_cores = np.array(dados)

matriz_cores = array_cores.reshape(8, 2)

print("\nMatriz de cores:")
print(matriz_cores)


nomes_pt = {
    "red": "vermelho",
    "green": "verde",
    "blue": "azul",
    "yellow": "amarelo",
    "cyan": "ciano",
    "magenta": "magenta",
    "black": "preto",
    "white": "branco"
}

for i in range(matriz_cores.shape[0]):
    matriz_cores[i, 0] = nomes_pt[matriz_cores[i, 0]]

print("\nMatriz com os nomes em português:")
print(matriz_cores)
```
