- 1. Carregue o Dataset paises.csv e em seguida mostre: a.Quais são os países da OCEANIA; b.Quantos países são da OCEANIA; Dica: para busca de padrões textuais no Pandas, use métodos da subclasse str da Series. Ex: series.str.contains(‘texto’)
- 2. Encontre o nome e a região do país que possui a maior população segundo este Dataset ;
- 3. Agrupe os países por Regiões. Em seguida, mostre a média de alfabetização (Literacy (%)) de cada região do planeta;
- 4. Busque o nome de todos os países do Dataset que não possuem costa marítima (Coastline (coast/area ratio) == 0) e guarde-os em um novo arquivo chamado noCoast.csv;
- 5. Faça uma função que receba a taxa de mortalidade de cada país (Deathrate) e retorne o texto ‘Balanced’ caso o valor seja < 9 e ‘Urgent’ caso contrário. Em seguida, crie um campo no Dataset chamado ‘Humanitarian Help’ que receba estes valores para cada país. No final, mostre o Dataset para verificar se a inserção da nova coluna foi feita com sucesso.

```python
import pandas as pd
import numpy as np

# Carregando o dataset
dataset = pd.read_csv('paises.csv', sep=';')


oceania = dataset[dataset['Region'].str.contains('OCEANIA', na=False)]

print("Países da OCEANIA:")
print(oceania['Country'])

print('\n')

print(f"Quantidade de países da OCEANIA: {len(oceania)}")
```

```python
maior = dataset['Population'].idxmax()

print("País com maior população:")
print(dataset.loc[maior, ['Country', 'Region']])
```

```python
media = dataset.groupby('Region')['Literacy (%)'].mean()

print("Média de alfabetização por região:")
print(media)
```

```python
semCosta = dataset[dataset['Coastline (coast/area ratio)'] == 0]

print("Países sem costa marítima:")
print(semCosta['Country'])

semCosta['Country'].to_csv('noCoast.csv', index=False)
```

```python
#pedi ajudo ao chat do colab, n consegui fazer a logica do exercico

#
def ajuda(valor):
    return 'Balanced' if valor < 9 else 'Urgent'
#

dataset['Humanitarian Help'] = dataset['Deathrate'].apply(ajuda)

print("Dataset com a nova coluna:")
print(dataset)
```

- 7. Crie uma função customizada que receba a coluna Infant mortality (per 1000 births ) de um país e retorne o valor reduzido em 15% (simulando assim uma meta de redução da mortalidade infantil). Aplique esta função à coluna usando o método apply() e, em seguida, concatene a coluna original com a coluna resultante lado a lado para se fazer uma comparação;
- 8. Remova a coluna Coastline (coast /area Ratio) do dataset. Em seguida, salve este novo dataset em um arquivo chamado paises_sem_coastline.csv

```python
def reducao(valor):
    return valor * 0.85

nova = dataset['Infant mortality (per 1000 births)'].apply(reducao)

comparacao = pd.concat(
    [dataset['Infant mortality (per 1000 births)'], nova],
    axis=1
)

print("Comparação da mortalidade infantil:")
print(comparacao)
```

```python
novoDataset = dataset.drop(columns=['Coastline (coast/area ratio)'])

novoDataset.to_csv('paises_sem_coastline.csv', index=False)

print("Dataset sem a coluna Coastline:")
print(novoDataset)
```
