TUPLAS (TUPLES) ()
1. guarda uma conjunto de elementos.
2. é uma coecao imutavel
3. mas permite o slicing dos seus elementos

```python
 nomes = ('Goku','Vegeta','Gohan','Trunks')
 print(nomes)
 print(type(nomes))

 # slicing de dados

 print(nomes)
 print(nomes[1])
 print(nomes[1:3])
 print(nomes[1:])
 print(nomes[-2])
```

LISTAS (LISTS) []

1. guarda um conjunto de elementos
2. coleção mutavel de dados

```python
 nomes = ['Goku','Vegeta','Gohan','Trunks']
 print(nomes)
 print(type(nomes))

#creat

 nomes.append('MaMajin Boo')
 print(nomes)

 nomes.insert(2,'Piccolo')
 print(nomes)

#read
#igual atuplas

#update
nomes[0]= 'tenshin han'
print(nomes)

#delete
del nomes[1]
print(nomes)
nomes.remove('Trunks')
print(nomes)
```

CONJUNTOS (SETS) {}

1. nao guarda ordem dos elelementos
2. nao guarda elementos repetidos
3. permite naturamente realizar operacoes entre conjunto de dados(uniao, deferença, interseçao...)

```python
nomes = {'Goku','Vegeta','Gohan','Trunks'}
print(nomes)
print(type(nomes))

#create
nomes.add('Piculo')

#read
print(nomes)
```

DICIONÁRIOS (DICTIONARIES)

1. organiza seus elementos
2. coleção mutável

```python
pessoa = {'nome':'Goku ',
            'idade':52,
            'sexo': 'M'
            }

print(pessoa)
print(type(pessoa))

##create
pessoa ['desenho'] = 'dbz'
print(pessoa)

##red

print(pessoa['nome'])

##update

pessoa['desenho'] = 'dbgt'

##delete

del pessoa['sexo']
print(pessoa)
```
